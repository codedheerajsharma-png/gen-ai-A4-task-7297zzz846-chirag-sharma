### task 4 generate summary report from file

```python
with open("sales_data.txt","r") as f:
    sale = [int(line.strip()) for line in f.readlines()]

print(f'sale is:{sale}')
print(f'total sale is: {sum(sale)}')
print(f'highest sale is: {max(sale)}')
print(f'lowest sale is: {min(sale)}')
print(f'average sale is: {sum(sale)/len(sale)}')
```

### task 5 create product info file (using input)

```python
list_1 = []

while True:
    product = lambda pro,pri:f'{pro} | {pri}\n'
    if len(list_1) == 3:
        break
    else:
        p = product(input('enter product name'),(input('enter price')))
        list_1.append(p)
```

```python

with open("products.txt","w") as f:
    f.writelines(list_1)
```

```python
with open("products.txt","r") as f:
    print(f.read())
```

```python
with open("products.txt", "r") as f:
    for line in f:
        product, price = line.strip().split(" | ")
        print(f"Product: {product}, Price: {price}")
```

### task 6 reading file safely (error handling inside file handling only)

```python
import os

filename = input("Enter filename: ")

if os.path.exists(filename):
    with open(filename,"r") as f:
        print(f.read())
else:
    print(f'file not found. check the filename.')
```

### task 7 mini project - export discounted prices

```python
prices = {"mouse":500,"keyboard":800,"monitor":7000,"pendrive":400,"camera":5000}

discount = float(input("enter discount in numeric value"))

total_d_price = 0

with open("discounted_report.txt","w") as f:
    f.write('product    |   price   |   discounted price\n')

    for i,v in prices.items():
        
        d = v-v*discount/100

        f.writelines(f'{i}      |   {v}     |   {d}\n')

        total_d_price += d

    average_d_price = d/len(prices)

    f.write(f'total of discounted prices is: {total_d_price}\n')
    f.write(f'average of discounted prices is: {average_d_price}\n')

with open("discounted_report.txt","r") as f:
    print(f.read())

```
